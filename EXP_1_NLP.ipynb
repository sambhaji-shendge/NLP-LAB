```python
import spacy
from nltk.stem import PorterStemmer

# Load spaCy
nlp = spacy.load("en_core_web_sm")

# Input sentence
text = "The students are studying natural language processing."

# Tokenization
doc = nlp(text)

print("Original Sentence:")
print(text)

print("\nTokens:")
for token in doc:
    print(token.text)

# Stemming
stemmer = PorterStemmer()

print("\nStemming:")
for token in doc:
    if token.is_alpha:
        print(token.text, "->", stemmer.stem(token.text))

# Lemmatization
print("\nLemmatization:")
for token in doc:
    if token.is_alpha:
        print(token.text, "->", token.lemma_)
```

# Experiment 1: Tokenization, Stemming and Lemmatization

## Aim
To implement tokenization, stemming and lemmatization using Python.

## Tools/Technologies Used

- NLTK
- spaCy

## Input Sentence
